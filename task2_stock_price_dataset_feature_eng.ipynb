```python
import pandas as pd
import numpy as np

# Load cleaned dataset
df = pd.read_csv("complete_stock_dataset_cleaned.csv")

# Ensure proper time order for time-series calculations
df = df.sort_values(["symbol", "date"])

# Feature engineering: daily return – percentage change in closing price
df["daily_return"] = df.groupby("symbol")["close price"].pct_change()

# Preview
df.head()
```

```python
# Feature engineering: daily return – percentage change in closing price
df["daily_return"] = df.groupby("symbol")["close price"].pct_change()

# Feature engineering: price range
df["price_range"] = df["high price"] - df["low price"]
```

```python
# Feature engineering: 10-day moving average
df["ma10"] = (
    df.groupby("symbol")["close price"]
      .transform(lambda x: x.rolling(10).mean())
)

# Feature engineering: 10-day rolling volatility
df["volatility_10"] = (
    df.groupby("symbol")["daily_return"]
      .transform(lambda x: x.rolling(10).std())
)
```

```python
# Removing rows with NaNs created by rolling features
df = df.dropna(subset=["daily_return", "ma10", "volatility_10"])
```
