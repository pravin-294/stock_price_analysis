```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df= pd.read_csv(r"C:\Users\Prajwal\OneDrive\Desktop\codveda data analysis task\data\raw\2) Stock Prices Data Set.csv")



df[df["symbol"] == "AAPL"].plot(x="date", y="close price", figsize=(12,5))
plt.show()

```

```python
df[df["symbol"] == "AAPL"].plot(x="date", y="volume traded", figsize=(12,5))
```

```python
df['daily_return'] = df.groupby('symbol')['close price'].pct_change()

df[['symbol', 'date', 'close price', 'daily_return']].head(20)
```

```python


df['daily_return'].hist(bins=50, figsize=(10,5))
plt.show()
```

```python
df['price_range'] = df['high price'] - df['low price']
df[['high price', 'low price', 'price_range']].head()
```

```python
df.boxplot(column="price_range", figsize=(10,5))
plt.show()
```
