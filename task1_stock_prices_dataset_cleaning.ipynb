```python
import pandas as pd
import numpy as np

df = pd.read_csv("Stock Prices Dataset raw.csv")

df.head()

#I converted the date column into datetime format for time-series analysis.”
df['date'] = pd.to_datetime(df['date'], format='%d-%m-%Y')
```

```python
# Basic data understanding and validation.

df.info()
df.describe()
df.isnull().sum()
df.duplicated().sum()
```

```python
df = df.sort_values(["symbol", "date"])
df[["open price", "high price", "low price"]] = (
    df.groupby("symbol")[["open price", "high price", "low price"]]
      .ffill()
      .bfill()
)

(df["low price"] > df["high price"]).sum()

```

```python
#Removing rows where low price was greater than high price.
df = df[df["low price"] <= df["high price"]]

#Keep only those rows where the low price is less than or equal to the high price. Remove everything else.

#Re-checked the dataset after cleaning.
(df["low price"] > df["high price"]).sum()
```

```python
# Display rows containing any remaining missing values (data validation check)

df[df.isnull().any(axis=1)]  
```

```python
# Checking for duplicate records based on symbol and date

df.duplicated(subset=["symbol", "date"]).sum()
```

```python
# Final data validation before saving cleaned dataset
print("Missing values:\n", df.isnull().sum())
print("\nDuplicate records (symbol + date):", 
      df.duplicated(subset=["symbol", "date"]).sum())

df.head()
```

```python
# Save dataset after data cleaning
df.to_csv("complete_stock_dataset_cleaned.csv", index=False)
```
