```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("complete_stock_dataset_cleaned.csv")


# Convert date column to datetime
df["date"] = pd.to_datetime(df["date"], dayfirst=True)


# Filter data for AAPL stock
aapl_df = df[df["symbol"] == "AAPL"]


# Sort data by date
aapl_df = aapl_df.sort_values("date")


# Plot closing price trend
plt.figure(figsize=(12, 5))
plt.plot(aapl_df["date"], aapl_df["close price"])
plt.title("AAPL Closing Price Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Closing Price")
plt.show()

# Line Plot (Time-Series Line Chart).
# AAPL is stocks ticker symbol of Apple Inc. 
```

```python
# AAPL Trading Volume over time.

# Convert daily trading volume into monthly total volume
monthly_volume = (
    aapl_df
    .set_index("date")["volume traded"]
    .resample("ME")
    .sum()
)

plt.figure(figsize=(12,5))
plt.bar(monthly_volume.index, monthly_volume.values, width=20)
plt.title("Monthly Trading Volume of AAPL")
plt.xlabel("Date")
plt.ylabel("Total Volume Traded")
plt.show()


# Monthly aggregation of AAPL trading volume for clearer visualization
# Bar chart showing monthly trading volume of AAPL stock.
```

```python
# Calculate daily percentage return for each stock symbol

df['daily_return'] = df.groupby('symbol')['close price'].pct_change()

df[['symbol', 'date', 'close price', 'daily_return']].head(20)


#NaN values are expected for the first trading day of each stock because there is no prior closing price to calculate the return.
```

```python

df['daily_return'].hist(bins=50, figsize=(10,5))
plt.title("Distribution of Daily Returns")
plt.xlabel("Daily Return")
plt.ylabel("Frequency")
plt.show()

# Histogram of daily returns for all stocks.
```

```python
#Difference between the highest and lowest price of a stock on a given day.

df['price_range'] = df['high price'] - df['low price']
df[['high price', 'low price', 'price_range']].head()
```

```python


df.boxplot(column="price_range", figsize=(10,5))
plt.title("Distribution of Daily Price Range")
plt.show()


#Box Plot (Box-and-Whisker Plot)
```
