```python
import pandas as pd
import numpy as np

df= pd.read_csv(r"C:\Users\Prajwal\OneDrive\Desktop\codveda data analysis task\data\raw\2) Stock Prices Data Set.csv")

df['daily_return'] = df.groupby('symbol')['close price'].pct_change()  #grouop by is used to divide the data based on each stock symbol before calculating the percentage change in closing prices.
df.head()

```

```python
df['price_range'] = df['high price'] - df['low price']
df.head(10)
```

```python
df['ma10'] = df.groupby('symbol')['close price'].transform(lambda x: x.rolling(10).mean())
df.head()
```

```python
df['volatility_10'] = df.groupby('symbol')['daily_return'].transform(lambda x: x.rolling(10).std())
df.head()
```
