```python
import pandas as pd
import numpy as np

df = pd.read_csv("complete_stock_dataset_cleaned.csv")
```

```python
df.info()
df.describe()
df.isnull().sum()
df.duplicated().sum()
```

```python
print(df.isnull().sum())   
#open,high, low having null values and this columns are numeric so we use mean to fill null values
```

```python
print(df["open price"].mean()) #to find the mean values of open column name "open"
print(df[["high price","low price"]].mean()) #to find the mean values of high price column and low price column 
```

```python
df["open price"]= df["open price"].replace(np.nan, 86.35)
print(df)
```

```python
df["high price"] = df["high price"].replace(np.nan, 87.13)
print(df)
```

```python
df["low price"] = df["low price"].replace(np.nan, 85.55)
print(df)
```

```python
df[df.isnull().any(axis=1)]  #shows only rows with missing values
```
